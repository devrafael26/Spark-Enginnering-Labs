```python
df = spark.range(1_000_000)

df_filtered = df.filter("id > 500000")

df_selected = df_filtered.select("id")

df_selected.show(5)
```

```python
df_selected.explain(True)
```

```python
# FLUXO

# spark.range()
#       ↓
# dados gerados pelo Spark
#       ↓
# filter()
#       ↓
# select()
#       ↓
# show()
#       ↓
# EXECUÇÃO
```
